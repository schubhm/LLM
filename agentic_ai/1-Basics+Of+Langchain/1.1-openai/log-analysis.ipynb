```python
import gzip
import os
import re
```

```python
# Set the path where your .log.gz files are stored
log_dir = "/Users/shubham.panday/Desktop/test"  # Change this to your actual log file directory
```

```python
# Regex pattern to match 'error' in log lines (case-insensitive)
error_pattern = re.compile(r'301', re.IGNORECASE)
```

```python
# Store error lines
error_lines = []
```

```python
# Scan all .log.gz files in the directory
for filename in os.listdir(log_dir):
    if filename.endswith(".log.gz"):
        filepath = os.path.join(log_dir, filename)
        print(f"Processing: {filename}")
        try:
            with gzip.open(filepath, 'rt', encoding='utf-8', errors='ignore') as file:
                for line in file:
                    if error_pattern.search(line):
                        error_lines.append((filename, line.strip()))
        except Exception as e:
            print(f"Failed to process {filename}: {e}")
```

```python
# Display the results
print(f"\n✅ Found {len(error_lines)} error lines across all files.\n")
```

```python
# Show the first 10 error lines
for i, (fname, line) in enumerate(error_lines[:10]):
    print(f"[{fname}] {line}")
```
